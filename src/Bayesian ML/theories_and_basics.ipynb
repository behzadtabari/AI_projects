## Basics
A random variable is a function that maps every possible outcome to a number, and the PDF/PMF takes that number and turns it into a probability. If summed upward, these PDFs create the cumulative distribution function (CDF). Below is a small toolkit that draws a binomial CDF and PDF. Feel free to change the probability of success and the total number of attempts.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# parameters (change freely)
n = 10      
p = 0.4
# x values
x = np.arange(0, n+1)

# PMF and CDF
pmf = binom.pmf(x, n, p)
cdf = binom.cdf(x, n, p)

# plot
plt.figure()

plt.stem(x, pmf, label="PMF (PDF for discrete)")
plt.step(x, cdf, where='post', label="CDF")

plt.xlabel("Number of successes")
plt.ylabel("Probability")
plt.title("Binomial Distribution: PMF and CDF")
plt.legend()

plt.show()
```
